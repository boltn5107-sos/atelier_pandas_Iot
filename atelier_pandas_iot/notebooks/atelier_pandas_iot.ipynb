### 5) Créer, de deux façons différentes, une Series de 4 températures avec les heures de mesure 12h, 13h, 14h et 15h comme index

**Méthode 1 :** passer l'index en paramètre

```python
temps = [24.5, 25.1, 26.3, 27.8]
heures = [12, 13, 14, 15]

serie_heures = pd.Series(temps, index=heures)
serie_heures
```

**Méthode 2 :** passer un dictionnaire

```python
serie_heures2 = pd.Series({12: 24.5, 13: 25.1, 14: 26.3, 15: 27.8})
serie_heures2
```

### 6) Afficher la 1ère et la dernière valeur

```python
premiere = serie_heures.iloc[0]
derniere = serie_heures.iloc[-1]
print(f"1ère valeur : {premiere}")
print(f"Dernière valeur : {derniere}")
```

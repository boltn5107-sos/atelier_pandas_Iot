## Partie 5 – Manipulation des colonnes


### 1) Ajouter la colonne "temperature_fahrenheit"

```python
df["temperature_fahrenheit"] = df["temperature"] * 9 / 5 + 32
df[["temperature", "temperature_fahrenheit"]].head()
```

### 2) Ajouter la colonne "niveau_temperature" (Élevée si température > 30, sinon Normale) avec la méthode where() de numpy

```python
import numpy as np

df["niveau_temperature"] = np.where(df["temperature"] > 30, "Élevée", "Normale")
df["niveau_temperature"].value_counts()
```

### 3) Renommer la colonne "humidite" en "humidite_relative"

```python
df = df.rename(columns={"humidite": "humidite_relative"})
df.columns
```

### 4) Supprimer la colonne "temperature_fahrenheit"

```python
df = df.drop(columns=["temperature_fahrenheit"])
df.columns
```

## Partie 6 – Filtrage


### 1) Lignes telles que température > 30

```python
df_filtre = df[df["temperature"] > 30]
print(f"{len(df_filtre)} lignes")
df_filtre[["temperature", "batiment"]].head()
```

### 2) Lignes telles que température > 30 et humidité > 70

```python
df_filtre2 = df[(df["temperature"] > 30) & (df["humidite_relative"] > 70)]
print(f"{len(df_filtre2)} lignes")
df_filtre2[["temperature", "humidite_relative", "batiment"]].head()
```

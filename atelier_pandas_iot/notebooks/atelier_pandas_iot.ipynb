## Partie 5 – Manipulation des colonnes


### 1) Ajouter la colonne "temperature_fahrenheit"

```python
df["temperature_fahrenheit"] = df["temperature"] * 9 / 5 + 32
df[["temperature", "temperature_fahrenheit"]].head()
```

### 2) Ajouter la colonne "niveau_temperature" (Élevée si température > 30, sinon Normale) avec la méthode where() de numpy

```python
import numpy as np

df["niveau_temperature"] = np.where(df["temperature"] > 30, "Élevée", "Normale")
df["niveau_temperature"].value_counts()
```

### 3) Renommer la colonne "humidite" en "humidite_relative"

```python
df = df.rename(columns={"humidite": "humidite_relative"})
df.columns
```

### 4) Supprimer la colonne "temperature_fahrenheit"

```python
df = df.drop(columns=["temperature_fahrenheit"])
df.columns
```

### 5) Remplacer les valeurs manquantes de température et d'humidité par les moyennes

```python
df["temperature"] = df["temperature"].fillna(df["temperature"].mean())
df["humidite_relative"] = df["humidite_relative"].fillna(df["humidite_relative"].mean())
df[["temperature", "humidite_relative"]].isna().sum()
```

### 6) Remplacer les valeurs manquantes de consommation par la médiane

```python
df["consommation"] = df["consommation"].fillna(df["consommation"].median())
df["consommation"].isna().sum()
```

### 7) Remplacer les valeurs manquantes de "etat" par "INCONNU"

```python
df["etat"] = df["etat"].fillna("INCONNU")
print(df["etat"].isna().sum(), "valeurs manquantes restantes")
df.isna().sum()
```

### 13.1 Analyse temporelle

```python
df["date_heure"] = pd.to_datetime(df["date_heure"])
df["jour"] = df["date_heure"].dt.day
df["heure"] = df["date_heure"].dt.hour
df[["date_heure", "jour", "heure", "temperature"]].head()
```

```python
temp_par_jour = df.groupby("jour")["temperature"].mean()
temp_par_jour
```

### 13.2 Détection d'anomalies (z-score)

```python
seuil = 2
moyenne = df["temperature"].mean()
ecart = df["temperature"].std()
df["anomalie_temp"] = (df["temperature"] - moyenne).abs() > seuil * ecart
print(f"Température moyenne : {moyenne:.2f}, écart-type : {ecart:.2f}")
print("Mesures anormales :", df["anomalie_temp"].sum())
df[df["anomalie_temp"]][["temperature", "batiment", "date_heure"]].head()
```

```python
df[df["anomalie_temp"]].groupby("batiment").size()
```

### 13.3 Matrice de corrélation

```python
corr = df[["temperature", "humidite_relative", "pression", "consommation"]].corr()
corr
```

### 13.4 Visualisations

```python
%matplotlib inline
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

temp_par_jour.plot(ax=axes[0], marker="o", color="tab:red", title="Température moyenne par jour")
df["etat"].value_counts().plot.bar(ax=axes[1], color="tab:green", title="Mesures par état")
df.groupby("batiment")["consommation"].mean().plot.bar(ax=axes[2], color="tab:blue", title="Consommation moyenne par bâtiment")

plt.tight_layout()
plt.show()
```

### 13.5 Consommation par heure

```python
conso_par_heure = df.groupby("heure")["consommation"].agg(["mean", "sum"])
conso_par_heure
```

```python
heure_pic = conso_par_heure["mean"].idxmax()
print(f"Le pic de consommation moyenne est atteint à {heure_pic}h")
```
